```python
# ASSIGNMENT 3
# Question 3
import random
roll_no = 1024170065
random.seed(roll_no)

nums = [random.randint(100, 900) for i in range(100)]

print("Random numbers:")
print(nums)

odd_nums = [i for i in nums if i % 2 != 0]

print("Odd numbers: ", odd_nums)
print("Number of odd numbers: ", len(odd_nums))

even_nums = [i for i in nums if i % 2 == 0]

print("Even numbers:", even_nums)
print("Number of even numbers:", len(even_nums))

def isPrime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


prime_nums = [i for i in nums if isPrime(i)]

print("Prime numbers:", prime_nums)
print("Number of prime numbers:", len(prime_nums))

freq = max(nums, key=nums.count)

print("Most frequent number:", freq)
print("It occurs: ", nums.count(freq), " times")
```
