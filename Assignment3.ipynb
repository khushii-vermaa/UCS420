```python
# QUESTION 6
employeesData = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 7, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}
df = pd.DataFrame(employeesData)
print(df)
```

```python
# a)
print(df.shape)
```

```python
# b)
df.info()
# c)
df.describe()
```

```python
# d)
print(df.head())
print(df.tail(3))
```

```python
# e)
print(f"i. Average salary: ${df['Salary'].mean()}")
print(f"ii. Total bonus paid: ${df['Bonus'].sum()}")
print(f"iii. Youngest employee's age: {df['Age'].min()} years")
print(f"iv. Highest performance rating: {df['Rating'].max()}")
# f)
sorted_df = df.sort_values(by='Salary', ascending=False)
print(sorted_df)
# g)
def categorize(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'
df['Performance_Category'] = df['Rating'].apply(categorize)
print(df[['Name', 'Rating', 'Performance_Category']])
# h)
print("\n")
print(df.isnull().sum())
print("Total missing values:", df.isnull().sum().sum())
# i)
df.rename(columns={'Employee_ID': 'ID'}, inplace=True)
print("DataFrame after renaming Employee_ID to ID:")
print(df.head())
# j)
print(df[df['Years_of_Experience'] > 5])
print("\n")
print(df[df['Department'] == 'IT'])
# k)
df['Tax'] = df['Salary'] * 0.10
print("DataFrame with Tax column:")
print(df[['Name', 'Salary', 'Tax']])
# l)
df.to_csv('modified.csv', index=False)
print("Modified DataFrame saved as 'modified.csv'")
```
