```python
#5.3
n = int(input("Enter n: "))
sum = 0
for num in range(2, n+1):
    prime = True
    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        sum = sum + num

print("Sum of prime numbers = ", sum)
```

```python
# 6.1
def addOdd(n):
  sum = 0
  for i in range(1, n+1):
    if i%2 != 0:
      sum = sum + i
  return sum

n = int(input("enter number: "))
print("Sum of odd numbers: ", addOdd(n))
# 6.2
def addPrime(n):
    sum = 0
    for i in range(2, n+1):
        prime = True
        for j in range(2, i):
            if i % j == 0:
                prime = False
                break

        if prime:
            sum = sum + i

    return sum

n = int(input("Enter n: "))
result = addPrime(n)
print("Sum of prime numbers =", result)
```
