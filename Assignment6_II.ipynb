```python
# QUESTION1
# Ask the user to enter a value (e.g., M) for a mathematical function.
#  Generate x values from -10 to 10 using np.linspace().
#  Compute y values for:
# o y= M⋅x2
# o y=M⋅sin(x)
#  Plot both functions in a single figure:
# o Use different colors and line styles.
# o Add legend, grid, and title.


M=float(input("Enter a number"))

x=np.linspace(-10,10,100)

y1=M*x**2
y2=M*np.sin(x)


plt.plot(x,y1,color='blue',linestyle='-',label='y=Mx^2')
plt.plot(x,y2,color='red',linestyle='--',label='y=Msin(x)')


plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")

plt.legend()
plt.grid(True)


plt.show()
```

```python
# Q.2 Create a dataset of five subjects and scores.
#  Convert it into a Pandas DataFrame.
#  Plot the scores using a Seaborn bar plot with:
# o Different colors for each bar.
# o Annotations on top of each bar.
# o Title, axis labels, and grid.



import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data={
    'Subjects':['Python','Operating systems','DBMS','Cognitive computing','Computer Science'],
    'Scores':[85,78,45,89,90]
}

df=pd.DataFrame(data)
print(df)

plt.figure(figsize=(10,6))
ax=sns.barplot(
    x='Subjects',
    y='Scores',
    data=df,
    hue='Subjects',
    palette='Set2',
    legend=False
)


for i in range(len(df)):
    ax.text(i, df['Scores'][i] + 1, str(df['Scores'][i]),
            ha='center')


plt.title("Subject-wise Scores")
plt.xlabel("Subjects")
plt.ylabel("Scores")

# Grid
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
# Q3. Write a Python script to select your roll number as seed for NumPy and
# then generate a dataset of 50 values using np.random.randn().
#  Create a 2x2 subplot layout displaying:
# o Line plot showing cumulative sum.
# o Scatter plot with random noise.
#  Customize with titles, axis labels, and grids.


import numpy as np
import matplotlib.pyplot as plt


roll_number = 1024170342
np.random.seed(roll_number)

data = np.random.randn(50)

x = np.arange(1, 51)

cumulative_sum = np.cumsum(data)

fig, ax = plt.subplots(2, 2, figsize=(12, 8))

ax[0, 0].plot(x, cumulative_sum)
ax[0, 0].set_title("Cumulative Sum")
ax[0, 0].set_xlabel("Index")
ax[0, 0].set_ylabel("Cumulative Sum")
ax[0, 0].grid(True)

noise = np.random.randn(50)

ax[0, 1].scatter(x, data + noise)
ax[0, 1].set_title("Scatter Plot with Random Noise")
ax[0, 1].set_xlabel("Index")
ax[0, 1].set_ylabel("Value")
ax[0, 1].grid(True)

ax[1, 0].plot(x, data)
ax[1, 0].set_title("Random Data")
ax[1, 0].set_xlabel("Index")
ax[1, 0].set_ylabel("Value")
ax[1, 0].grid(True)

ax[1, 1].hist(data, bins=10)
ax[1, 1].set_title("Distribution of Random Data")
ax[1, 1].set_xlabel("Value")
ax[1, 1].set_ylabel("Frequency")
ax[1, 1].grid(True)

plt.tight_layout()


plt.show()
```

```python
# Q.4 Download Data-set from the below link
# https://github.com/AnjulaMehto/MCA/blob/main/company_sales_data.csv
# Apply ‘seaborn’ library to do the following.
# 1. Read Total profit of all months and show it using a line plot.
# 2. Read all product sales data and show it using a multiline plot.
# 3. Plot bar chart for all the features/attributes.


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"

df = pd.read_csv(url)

print(df)

plt.figure(figsize=(10, 6))

sns.lineplot(
    x='month_number',
    y='total_profit',
    data=df,
    marker='o'
)

plt.title("Total Profit of All Months")
plt.xlabel("Month Number")
plt.ylabel("Total Profit")

plt.grid(True)

plt.show()



products = [
    'facecream',
    'facewash',
    'toothpaste',
    'bathingsoap',
    'shampoo',
    'moisturizer'
]

plt.figure(figsize=(12, 7))

for product in products:
    sns.lineplot(
        x=df['month_number'],
        y=df[product],
        marker='o',
        label=product
    )

plt.title("Monthly Product Sales")
plt.xlabel("Month Number")
plt.ylabel("Units Sold")

plt.legend()
plt.grid(True)

plt.show()




products = [
    'facecream',
    'facewash',
    'toothpaste',
    'bathingsoap',
    'shampoo',
    'moisturizer'
]

df_melted = df.melt(
    id_vars='month_number',
    value_vars=products,
    var_name='Product',
    value_name='Sales'
)

plt.figure(figsize=(14, 7))

sns.barplot(
    x='month_number',
    y='Sales',
    hue='Product',
    data=df_melted
)

plt.title("Product Sales for All Months")
plt.xlabel("Month Number")
plt.ylabel("Sales")

plt.grid(axis='y')

plt.show()
```
